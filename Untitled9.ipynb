```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
import pandas as pd
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# 1. Load Data
print("Loading datasets...")
energy_df = pd.read_csv("energy_dataset.csv")
weather_df = pd.read_csv("weather_features.csv")

# 2. Data Cleaning & Preprocessing
energy_df["time"] = pd.to_datetime(energy_df["time"], utc=True)
weather_df["dt_iso"] = pd.to_datetime(weather_df["dt_iso"], utc=True)

# Clean column names in energy dataframe for convenience
energy_df.columns = energy_df.columns.str.strip().str.replace(" ", "_")

# Handle missing values using interpolation
energy_df = energy_df.interpolate(method="linear").bfill().ffill()
weather_df = weather_df.interpolate(method="linear").bfill().ffill()

# 3. Aggregating Weather Features across Cities (Hourly Averages)
weather_agg = weather_df.groupby("dt_iso").agg({"temp": "mean", "pressure": "mean", "humidity": "mean", "wind_speed": "mean", "clouds_all": "mean"}).reset_index()

# Merge Datasets on Timestamp
df = pd.merge(energy_df, weather_agg, left_on="time", right_on="dt_iso", how="inner")
df["hour"] = df["time"].dt.hour
df["month"] = df["time"].dt.month

print(f"Merged dataset shape: {df.shape}")

# 4. Advanced Feature Engineering (Equivalent to SQL Window Functions)
df["price_rolling_7d"] = df["price_actual"].rolling(window=168).mean()  # 168 hours = 7 days
df["load_lag_1h"] = df["total_load_actual"].shift(1)
df["load_ramping_rate"] = df["total_load_actual"] - df["load_lag_1h"]

# 5. Statistical Summary & Correlation
print("\n--- Statistical Summary ---")
print(df[["total_load_actual", "price_actual", "temp", "wind_speed"]].describe())

plt.figure(figsize=(10, 8))
corr_cols = ["total_load_actual", "price_actual", "temp", "wind_speed", "generation_solar", "generation_wind_onshore", "generation_nuclear"]
corr_matrix = df[corr_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Matrix - Spain Energy & Weather")
plt.tight_layout()
plt.savefig("correlation_heatmap.png")
plt.close()

# 6. 2D Visualizations
# Stacked Area Chart for Energy Generation Mix
plt.figure(figsize=(14, 6))
sample_df = df.iloc[::24]  # Sample daily to keep plot readable
plt.stackplot(
    sample_df["time"],
    sample_df["generation_nuclear"].fillna(0),
    sample_df["generation_fossil_gas"].fillna(0),
    sample_df["generation_wind_onshore"].fillna(0),
    sample_df["generation_solar"].fillna(0),
    labels=["Nuclear", "Fossil Gas", "Wind Onshore", "Solar"],
)
plt.legend(loc="upper left")
plt.title("Daily Generation Mix Over Time")
plt.xlabel("Time")
plt.ylabel("Generation (MW)")
plt.tight_layout()
plt.savefig("generation_mix_stacked.png")
plt.close()

# 7. 3D Graphical Representations
# 3D Scatter Plot: Wind Speed vs Temperature vs Actual Price
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection="3d")

# Subsample for cleaner 3D render
sub_3d = df.sample(n=2000, random_state=42)
xs = sub_3d["wind_speed"]
ys = sub_3d["temp"]
zs = sub_3d["price_actual"]
cs = sub_3d["total_load_actual"]

img = ax.scatter(xs, ys, zs, c=cs, cmap="viridis", alpha=0.6, s=20)
fig.colorbar(img, label="Total Load Actual (MW)")
ax.set_xlabel("Wind Speed (m/s)")
ax.set_ylabel("Temperature (K)")
ax.set_zlabel("Price Actual (€/MWh)")
ax.set_title("3D Scatter: Wind Speed vs Temperature vs Energy Price")
plt.savefig("3d_scatter_analysis.png")
plt.close()

# 3D Surface Plot: Hour of Day vs Month vs Total Load
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection="3d")

pivot_table = df.pivot_table(values="total_load_actual", index="month", columns="hour", aggfunc="mean")
X = pivot_table.columns.to_numpy()
Y = pivot_table.index.to_numpy()
X, Y = np.meshgrid(X, Y)
Z = pivot_table.to_numpy()

surf = ax.plot_surface(X, Y, Z, cmap="plasma", edgecolor="none", alpha=0.9)
fig.colorbar(surf, shrink=0.5, aspect=5, label="Mean Load (MW)")
ax.set_xlabel("Hour of Day")
ax.set_ylabel("Month")
ax.set_zlabel("Actual Total Load (MW)")
ax.set_title("3D Surface Plot: Consumption Patterns across Hours and Months")
plt.savefig("3d_surface_consumption.png")
plt.close()

print("\nAll advanced queries, features, and visualizations completed successfully!")
print("Generated artifacts: 'correlation_heatmap.png', 'generation_mix_stacked.png', '3d_scatter_analysis.png', '3d_surface_consumption.png'")
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
import pandas as pd
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# 1. Load Data
print("Loading datasets...")
energy_df = pd.read_csv("energy_dataset.csv")
weather_df = pd.read_csv("weather_features.csv")

# 2. Data Cleaning & Preprocessing
energy_df["time"] = pd.to_datetime(energy_df["time"], utc=True)
weather_df["dt_iso"] = pd.to_datetime(weather_df["dt_iso"], utc=True)

# Clean column names in energy dataframe for convenience
energy_df.columns = energy_df.columns.str.strip().str.replace(" ", "_")
weather_df["city_name"] = weather_df["city_name"].str.strip()

# Handle missing values using interpolation
energy_df = energy_df.interpolate(method="linear").bfill().ffill()
weather_df = weather_df.interpolate(method="linear").bfill().ffill()

# 3. Advanced Multi-City Weather Aggregation (Spatial Extremes & Means)
# Extracting city-specific metrics or regional min/max to capture nationwide extremes
weather_agg = (
    weather_df.groupby("dt_iso")
    .agg(
        {
            "temp": ["mean", "min", "max"],
            "pressure": "mean",
            "humidity": "mean",
            "wind_speed": ["mean", "max"],
            "clouds_all": "mean",
            "rain_1h": "sum",
        }
    )
    .reset_index()
)

# Flatten multi-index columns
weather_agg.columns = ["_".join(col).strip("_") for col in weather_agg.columns.values]
weather_agg = weather_agg.rename(columns={"dt_iso_": "dt_iso"})

# Merge Datasets on Timestamp
df = pd.merge(energy_df, weather_agg, left_on="time", right_on="dt_iso", how="inner")
df["hour"] = df["time"].dt.hour
df["month"] = df["time"].dt.month
df["day_of_week"] = df["time"].dt.dayofweek
df["is_weekend"] = df["day_of_week"].apply(lambda x: 1 if x >= 5 else 0)

print(f"Merged dataset shape with advanced spatial metrics: {df.shape}")

# 4. Advanced Feature Engineering & Time-Series Modeling
# Rolling Market Volatility (7-day standard deviation of actual prices)
df["price_volatility_7d"] = df["price_actual"].rolling(window=168).std()

# Forecast Error Analysis (Actual vs Forecast Load)
df["load_forecast_error"] = df["total_load_actual"] - df["total_load_forecast"]
df["load_forecast_error_pct"] = (df["load_forecast_error"] / df["total_load_actual"]) * 100

# Ramping Rates (Hourly change)
df["load_ramping_rate"] = df["total_load_actual"].diff()

# Renewable Capacity Factor Approximations (assuming peak installs for normalization)
df["wind_capacity_factor"] = df["generation_wind_onshore"] / df["generation_wind_onshore"].max()
df["solar_capacity_factor"] = df["generation_solar"] / df["generation_solar"].max()

# 5. Advanced Statistical Distributions & Violin Plots
plt.figure(figsize=(14, 6))
sns.violinplot(
    data=df,
    x="month",
    y="price_actual",
    palette="viridis",
    inner="quartile",
)
plt.title("Monthly Distribution & Volatility of Actual Energy Prices (€/MWh)")
plt.xlabel("Month")
plt.ylabel("Price Actual (€/MWh)")
plt.tight_layout()
plt.savefig("advanced_price_violin.png")
plt.close()

# 6. Advanced 2D Subplots: Forecast Errors & Ramping Rates
fig, axes = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

axes[0].plot(
    df["time"].iloc[::24],
    df["load_forecast_error"].iloc[::24],
    color="crimson",
    alpha=0.7,
    label="Load Forecast Error (Actual - Forecast)",
)
axes[0].axhline(0, color="black", linestyle="--", linewidth=1)
axes[0].set_title("Daily Total Load Forecast Residuals (MW)")
axes[0].set_ylabel("Error (MW)")
axes[0].legend(loc="upper left")

axes[1].plot(
    df["time"].iloc[::24],
    df["price_volatility_7d"].iloc[::24],
    color="darkorange",
    label="7-Day Rolling Price Volatility (Std Dev)",
)
axes[1].set_title("Market Price Volatility Trend Over Time")
axes[1].set_xlabel("Time")
axes[1].set_ylabel("Volatility (€/MWh)")
axes[1].legend(loc="upper left")

plt.tight_layout()
plt.savefig("advanced_residuals_volatility.png")
plt.close()

# 7. Advanced 3D Interactive-Style Representation: Wind Speed vs Solar vs Price
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection="3d")

sub_3d = df.sample(n=3000, random_state=42)
xs = sub_3d["wind_speed_mean"]
ys = sub_3d["generation_solar"].fillna(0)
zs = sub_3d["price_actual"]
cs = sub_3d["temp_mean"]

img = ax.scatter(xs, ys, zs, c=cs, cmap="coolwarm", alpha=0.7, s=25)
fig.colorbar(img, label="Mean Temperature (K)")
ax.set_xlabel("Mean Wind Speed (m/s)")
ax.set_ylabel("Solar Generation (MW)")
ax.set_zlabel("Price Actual (€/MWh)")
ax.set_title("Advanced 3D Analysis: Wind, Solar Generation, and Pricing Dynamics")
plt.savefig("advanced_3d_multivar.png")
plt.close()

print("\nAdvanced modeling, error tracking, and visualizations completed successfully!")
print("New artifacts generated: 'advanced_price_violin.png', 'advanced_residuals_volatility.png', 'advanced_3d_multivar.png'")
```
