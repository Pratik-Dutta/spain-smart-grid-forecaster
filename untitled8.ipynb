```python
class SmartGridMetaLearner:
    """
    Blends Tabular Machine Learning (LightGBM) with Time-Series LLMs
    (Chronos) using an optimized Ridge Meta-Learner model.
    """
    def __init__(self):
        # Constrain to positive alpha weights to prevent inverse tracking anomalies
        self.meta_model = Ridge(alpha=1.0, positive=True, fit_intercept=False)

    def train_meta_learner(self, calibration_df, target_col):
        """
        Learns the operational strengths of both models from validation data.
        """
        print("🔗 Training Stacking Meta-Learner on base model residual vectors...")
        # Extract individual model prediction fields
        X_calib = calibration_df[['lgb_pred', 'chronos_pred']].values
        y_calib = calibration_df[target_col].values

        self.meta_model.fit(X_calib, y_calib)

        # Save learned weight coefficients
        self.lgb_weight = self.meta_model.coef_[0]
        self.chronos_weight = self.meta_model.coef_[1]

        print(f"📊 Learned Blending Architecture Weights -> LightGBM: {self.lgb_weight:.3f} | Chronos: {self.chronos_weight:.3f}")

    def predict_hybrid(self, inference_df):
        """Generates the optimal stacked prediction vector."""
        X_inf = inference_df[['lgb_pred', 'chronos_pred']].values
        return self.meta_model.predict(X_inf)
```

```python
import pandas as pd
import numpy as np
import logging
from pathlib import Path
from typing import Optional

# --- Logging setup (cleaner than scattered print statements) ---
logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
logger = logging.getLogger("energy_pipeline")


def load_csv_safely(path: str, required_cols: Optional[list] = None) -> pd.DataFrame:
    """Load a CSV with existence, emptiness, and schema validation."""
    file_path = Path(path)
    if not file_path.exists():
        raise FileNotFoundError(f"File not found: {path}")

    df = pd.read_csv(file_path)

    if df.empty:
        raise ValueError(f"{path} loaded but contains no rows.")

    if required_cols:
        missing = set(required_cols) - set(df.columns)
        if missing:
            raise KeyError(f"{path} is missing expected columns: {missing}")

    logger.info(f"Loaded {path} → shape={df.shape}")
    return df


def parse_datetime_column(df: pd.DataFrame, col: str, utc: bool = True) -> pd.DataFrame:
    """Convert a column to datetime, coercing bad values to NaT and reporting them."""
    before_na = df[col].isna().sum()
    df[col] = pd.to_datetime(df[col], utc=utc, errors="coerce")
    after_na = df[col].isna().sum()

    dropped = after_na - before_na
    if dropped > 0:
        logger.warning(f"{dropped} unparseable timestamps in '{col}' converted to NaT.")

    return df


def optimize_dtypes(df: pd.DataFrame) -> pd.DataFrame:
    """Downcast numeric columns to save memory — useful for large weather/energy datasets."""
    float_cols = df.select_dtypes(include=["float64"]).columns
    int_cols = df.select_dtypes(include=["int64"]).columns

    df[float_cols] = df[float_cols].apply(pd.to_numeric, downcast="float")
    df[int_cols] = df[int_cols].apply(pd.to_numeric, downcast="integer")

    return df


def build_weather_aggregate(
    df_weather: pd.DataFrame,
    dt_col: str = "dt_iso",
    agg_map: Optional[dict] = None,
) -> pd.DataFrame:
    """
    Aggregate multi-city weather data into a single national signal per timestamp.
    Defaults to mean, but agg_map lets you customize per-column logic
    (e.g. max wind_speed for grid-risk modeling instead of mean).
    """
    agg_map = agg_map or {"temp": "mean", "humidity": "mean", "wind_speed": "mean"}

    missing = set(agg_map.keys()) - set(df_weather.columns)
    if missing:
        raise KeyError(f"Weather aggregation columns missing from data: {missing}")

    df_agg = df_weather.groupby(dt_col).agg(agg_map).reset_index()
    logger.info(f"Aggregated weather data → {df_agg.shape[0]} unique timestamps.")
    return df_agg


def merge_energy_weather(
    df_energy: pd.DataFrame,
    df_weather_clean: pd.DataFrame,
    energy_time_col: str = "time",
    weather_time_col: str = "dt_iso",
) -> pd.DataFrame:
    """Merge energy and weather data with pre/post row-count diagnostics."""
    pre_rows = len(df_energy)

    df = pd.merge(
        df_energy,
        df_weather_clean,
        left_on=energy_time_col,
        right_on=weather_time_col,
        how="inner",
    )

    post_rows = len(df)
    dropped_pct = (1 - post_rows / pre_rows) * 100 if pre_rows else 0

    logger.info(f"Merged dataset shape: {df.shape}")
    if dropped_pct > 0:
        logger.warning(
            f"{dropped_pct:.2f}% of energy rows had no matching weather timestamp and were dropped."
        )

    return df



ENERGY_CSV = "energy_dataset.csv"
WEATHER_CSV = "weather_features.csv"

df_energy = load_csv_safely(ENERGY_CSV, required_cols=["time"])
df_weather = load_csv_safely(WEATHER_CSV, required_cols=["dt_iso", "temp", "humidity", "wind_speed"])

df_energy = parse_datetime_column(df_energy, "time")
df_weather = parse_datetime_column(df_weather, "dt_iso")

df_weather_clean = build_weather_aggregate(df_weather)

df = merge_energy_weather(df_energy, df_weather_clean)
print(df.columns.tolist())  # sanity check

df = df.set_index("time").sort_index()
df = optimize_dtypes(df)

logger.info(f"Final dataset ready → shape={df.shape}, memory={df.memory_usage(deep=True).sum() / 1e6:.2f} MB")
```

```python
import warnings
warnings.filterwarnings("ignore")

import logging
from pathlib import Path
from typing import Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import lightgbm as lgb
import shap
from statsmodels.tsa.seasonal import seasonal_decompose
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

import torch
from chronos import BaseChronosPipeline

logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
logger = logging.getLogger("energy_pipeline")
```

```python
def load_csv_safely(path: str, required_cols: Optional[list] = None) -> pd.DataFrame:
    """Load a CSV with existence, emptiness, and schema validation."""
    file_path = Path(path)
    if not file_path.exists():
        raise FileNotFoundError(f"File not found: {path}")

    df = pd.read_csv(file_path)
    if df.empty:
        raise ValueError(f"{path} loaded but contains no rows.")

    if required_cols:
        missing = set(required_cols) - set(df.columns)
        if missing:
            raise KeyError(f"{path} is missing expected columns: {missing}")

    logger.info(f"Loaded {path} -> shape={df.shape}")
    return df


def detect_target_column(df: pd.DataFrame) -> str:
    """Auto-detect the real-world 'total load actual' column regardless of
    whether it uses spaces or underscores in the source file."""
    candidates = [
        c for c in df.columns
        if "total" in c.lower() and "load" in c.lower() and "actual" in c.lower()
    ]
    if not candidates:
        raise KeyError(
            f"Could not locate a 'total load actual' style column. "
            f"Available columns: {df.columns.tolist()}"
        )
    return candidates[0]
```

```python
CITY_POPULATION_WEIGHTS = {
    "Madrid": 6_700_000,
    "Barcelona": 5_500_000,
    "Valencia": 1_650_000,
    "Seville": 1_300_000,
    "Bilbao": 1_000_000,
}


def build_population_weighted_weather(df_weather: pd.DataFrame) -> pd.DataFrame:
    """Collapse 5-city weather data into a single population-weighted
    national signal per timestamp, instead of a naive city-average."""
    df_weather = df_weather.copy()
    df_weather["city_name"] = df_weather["city_name"].str.strip()

    unknown_cities = set(df_weather["city_name"].unique()) - set(CITY_POPULATION_WEIGHTS)
    if unknown_cities:
        logger.warning(f"Unweighted / unknown cities found, dropping: {unknown_cities}")
        df_weather = df_weather[df_weather["city_name"].isin(CITY_POPULATION_WEIGHTS)]

    df_weather["weight"] = df_weather["city_name"].map(CITY_POPULATION_WEIGHTS)

    def weighted_avg(group: pd.DataFrame, col: str) -> float:
        return np.average(group[col], weights=group["weight"])

    records = []
    for dt, group in df_weather.groupby("dt_iso"):
        records.append({
            "dt_iso": dt,
            "temp": weighted_avg(group, "temp"),
            "humidity": weighted_avg(group, "humidity"),
            "wind_speed": weighted_avg(group, "wind_speed"),
        })
    df_agg = pd.DataFrame(records)
    logger.info(f"Population-weighted weather aggregated -> {df_agg.shape[0]} timestamps.")
    return df_agg
```

```python
def add_fourier_terms(df: pd.DataFrame, period: int, order: int, prefix: str) -> pd.DataFrame:
    """Generic Fourier-term generator. `order` harmonics capture richer
    seasonal shape than a single sin/cos pair (e.g. double-peak daily
    industrial demand curves)."""
    t = np.arange(len(df))
    for k in range(1, order + 1):
        df[f"{prefix}_sin_{k}"] = np.sin(2 * np.pi * k * t / period)
        df[f"{prefix}_cos_{k}"] = np.cos(2 * np.pi * k * t / period)
    return df


def load_and_engineer_data(energy_path: str, weather_path: str):
    print("🧹 Parsing and aligning multi-city industrial data vectors...")

    df_energy = load_csv_safely(energy_path, required_cols=["time"])
    df_weather = load_csv_safely(
        weather_path, required_cols=["dt_iso", "city_name", "temp", "humidity", "wind_speed"]
    )

    # --- Time-zone alignment: Spain's TSO data is timezone-aware; normalize
    # everything to UTC to avoid silent misalignment across DST transitions ---
    df_energy["time"] = pd.to_datetime(df_energy["time"], utc=True)
    df_weather["dt_iso"] = pd.to_datetime(df_weather["dt_iso"], utc=True)

    df_weather_clean = build_population_weighted_weather(df_weather)

    df = pd.merge(df_energy, df_weather_clean, left_on="time", right_on="dt_iso", how="inner")
    dropped_pct = (1 - len(df) / len(df_energy)) * 100
    if dropped_pct > 0:
        logger.warning(f"{dropped_pct:.2f}% of energy rows dropped on weather merge.")

    df = df.set_index("time").sort_index()

    target_col = detect_target_column(df)
    print(f"✅ Detected target column: '{target_col}'")
    df[target_col] = df[target_col].ffill()

    # --- Keep the TSO's own official day-ahead forecast as a benchmark
    # column: this lets us later ask "did our model beat the grid operator's
    # own forecast?" instead of only benchmarking against a naive baseline ---
    if "total load forecast" in df.columns:
        df["tso_forecast_benchmark"] = df["total load forecast"]

    # Scale temperature metric to Celsius if required
    df["temp_c"] = df["temp"] - 273.15 if df["temp"].mean() > 200 else df["temp"]

    # Industrial Degree Days
    df["CDD"] = np.maximum(0, df["temp_c"] - 22.0)
    df["HDD"] = np.maximum(0, 18.0 - df["temp_c"])

    # Calendar-cyclic signals (kept for LightGBM interpretability via SHAP)
    df["hour"] = df.index.hour
    df["day_of_week"] = df.index.dayofweek
    df["month"] = df.index.month
    df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24.0)
    df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24.0)
    df["day_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7.0)
    df["day_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7.0)

    # Higher-order Fourier terms for annual seasonality (captures summer AC
    # peaks + winter heating peaks better than a single sinusoid)
    df = add_fourier_terms(df, period=365.25 * 24, order=3, prefix="annual")

    # Structural autoregressive lags
    df["load_lag_24"] = df[target_col].shift(24)
    df["load_lag_168"] = df[target_col].shift(168)
    df["load_rolling_mean_24"] = df[target_col].shift(1).rolling(window=24).mean()
    df["load_rolling_std_24"] = df[target_col].shift(1).rolling(window=24).std()

    return df.dropna(), target_col
```

```python
def load_and_engineer_data(energy_path, weather_path):
    print("🧹 Parsing and aligning multi-city industrial data vectors...")

    df_energy = load_csv_safely(energy_path, required_cols=["time"])
    df_weather = load_csv_safely(
        weather_path, required_cols=["dt_iso", "city_name", "temp", "humidity", "wind_speed"]
    )

    df_energy["time"] = pd.to_datetime(df_energy["time"], utc=True)
    df_weather["dt_iso"] = pd.to_datetime(df_weather["dt_iso"], utc=True)

    df_weather_clean = build_population_weighted_weather(df_weather)

    df = pd.merge(df_energy, df_weather_clean, left_on="time", right_on="dt_iso", how="inner")
    df = df.set_index("time").sort_index()

    target_col = detect_target_column(df)
    print(f"✅ Detected target column: '{target_col}'")
    df[target_col] = df[target_col].ffill()

    if "total load forecast" in df.columns:
        df["tso_forecast_benchmark"] = df["total load forecast"]

    df["temp_c"] = df["temp"] - 273.15 if df["temp"].mean() > 200 else df["temp"]
    df["CDD"] = np.maximum(0, df["temp_c"] - 22.0)
    df["HDD"] = np.maximum(0, 18.0 - df["temp_c"])

    df["hour"] = df.index.hour
    df["day_of_week"] = df.index.dayofweek
    df["month"] = df.index.month
    df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24.0)
    df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24.0)
    df["day_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7.0)
    df["day_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7.0)

    df = add_fourier_terms(df, period=365.25 * 24, order=3, prefix="annual")

    df["load_lag_24"] = df[target_col].shift(24)
    df["load_lag_168"] = df[target_col].shift(168)
    df["load_rolling_mean_24"] = df[target_col].shift(1).rolling(window=24).mean()
    df["load_rolling_std_24"] = df[target_col].shift(1).rolling(window=24).std()

    # ---- THE FIX: only require non-null values in columns this pipeline
    # actually needs, instead of df.dropna() which wipes every row because
    # some unused raw columns (e.g. 'generation hydro pumped storage
    # aggregated') are 100% NaN across the whole dataset ----
    required_cols = [
        target_col, "temp_c", "CDD", "HDD",
        "hour_sin", "hour_cos", "day_sin", "day_cos",
        "load_lag_24", "load_lag_168", "load_rolling_mean_24", "load_rolling_std_24",
    ] + [c for c in df.columns if c.startswith("annual_")]

    if "tso_forecast_benchmark" in df.columns:
        required_cols.append("tso_forecast_benchmark")

    df_clean = df.dropna(subset=required_cols)
    return df_clean, target_col
```

```python
forecast_horizon = 24

fourier_cols = [c for c in processed_df.columns if c.startswith("annual_")]
features = [
    "temp_c", "humidity", "wind_speed", "CDD", "HDD",
    "hour_sin", "hour_cos", "day_sin", "day_cos",
    "load_lag_24", "load_lag_168", "load_rolling_mean_24", "load_rolling_std_24",
] + fourier_cols

missing_feats = [f for f in features if f not in processed_df.columns]
if missing_feats:
    raise KeyError(f"Missing expected feature columns: {missing_feats}")

train_df = processed_df.iloc[:-forecast_horizon]
test_df = processed_df.iloc[-forecast_horizon:].copy()

model = lgb.LGBMRegressor(
    n_estimators=300,
    learning_rate=0.05,
    num_leaves=31,
    random_state=42,
    verbosity=-1,
)
model.fit(train_df[features], train_df[target])
test_df["lgb_pred"] = model.predict(test_df[features])

lgb_mae = mean_absolute_error(test_df[target], test_df["lgb_pred"])
lgb_rmse = root_mean_squared_error(test_df[target], test_df["lgb_pred"])
print(f"LightGBM  -> MAE: {lgb_mae:,.1f} MW | RMSE: {lgb_rmse:,.1f} MW")

if "tso_forecast_benchmark" in test_df.columns:
    tso_mae = mean_absolute_error(test_df[target], test_df["tso_forecast_benchmark"])
    tso_rmse = root_mean_squared_error(test_df[target], test_df["tso_forecast_benchmark"])
    print(f"TSO Forecast -> MAE: {tso_mae:,.1f} MW | RMSE: {tso_rmse:,.1f} MW")
```

```python
print("\n📊 Calculating model transparency matrix using SHAP...")
explainer = shap.TreeExplainer(model)
sample_size = min(2000, len(processed_df))
shap_sample = processed_df[features].sample(sample_size, random_state=42)
shap_values = explainer(shap_sample)

plt.figure(figsize=(10, 5))
shap.summary_plot(shap_values, shap_sample, show=False)
plt.title("SHAP Feature Attribution: Drivers of Smart-Grid Energy Fluctuations", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"⚙️  Loading Chronos-Bolt foundation model on device: {device}")

chronos_pipeline = BaseChronosPipeline.from_pretrained(
    "amazon/chronos-bolt-base",
    device_map=device,
    torch_dtype=torch.bfloat16 if device == "cuda" else torch.float32,
)

context_series = torch.tensor(train_df[target].values, dtype=torch.float32)

quantiles, mean_forecast = chronos_pipeline.predict_quantiles(
    inputs=context_series,
    prediction_length=forecast_horizon,
    quantile_levels=[0.1, 0.5, 0.9],
)

test_df["chronos_pred"] = mean_forecast[0].numpy()
test_df["chronos_lower_10"] = quantiles[0, :, 0].numpy()
test_df["chronos_upper_90"] = quantiles[0, :, 2].numpy()

chronos_mae = mean_absolute_error(test_df[target], test_df["chronos_pred"])
chronos_rmse = root_mean_squared_error(test_df[target], test_df["chronos_pred"])
print(f"Chronos (zero-shot) -> MAE: {chronos_mae:,.1f} MW | RMSE: {chronos_rmse:,.1f} MW")
```

```python
def execute_advanced_layer(test_df, target_col):
    print("🚀 Injecting Advanced Hybrid Orchestration Layer...")

    # Instantiate our blending orchestrator class
    orchestrator = SmartGridMetaLearner()

    # Split the evaluation window in half to prevent data leakage
    midpoint = len(test_df) // 2
    calibration_slice = test_df.iloc[:midpoint].copy()
    evaluation_slice = test_df.iloc[midpoint:].copy()

    # Fallback safety guard if data rows are minimal
    if len(calibration_slice) == 0 or len(evaluation_slice) == 0:
        calibration_slice = test_df.copy()
        evaluation_slice = test_df.copy()

    # Fit the meta-regressor weights over the calibration split
    orchestrator.train_meta_learner(calibration_slice, target_col)

    # Compute the final advanced combined forecast array
    evaluation_slice['hybrid_ensemble_pred'] = orchestrator.predict_hybrid(evaluation_slice)

    return evaluation_slice, orchestrator

# Execute the partition split
advanced_eval_df, trained_orchestrator = execute_advanced_layer(test_df, target)
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

print("📊 Advanced statistical diagnostic packages loaded successfully.")
```

```python
# 1. Clean the variable strings to comply with statsmodels formula formatting constraints
stat_df = processed_df.copy()
stat_df = stat_df.rename(columns={target: 'target_load'})

# 2. Divide into historical training context and validation slices
forecast_horizon = 24
train_stat = stat_df.iloc[:-forecast_horizon]
test_stat = stat_df.iloc[-forecast_horizon:].copy()

# 3. Formulate the structural regression equation containing weather and autoregressive features
formula = "target_load ~ temp_c + CDD + HDD + hour_sin + hour_cos + load_lag_24"

print("📐 Fitting Multi-Quantile Time-Series Framework...")
# Fit Median Load Model (Quantile 0.50)
median_model = smf.quantreg(formula, data=train_stat).fit(q=0.50)
# Fit Extreme Peak Risk Model (Quantile 0.95)
peak_model = smf.quantreg(formula, data=train_stat).fit(q=0.95)

# 4. Generate out-of-sample statistical predictions
test_stat['stat_median_pred'] = median_model.predict(test_stat)
test_stat['stat_peak_pred'] = peak_model.predict(test_stat)

print("✅ Quantile estimation processing complete.")
```

```python
# Extract parameters for the 95th Percentile Model to show impact during extreme spikes
summary_coefs = peak_model.summary2().tables[1]

# Format the dataframe for presentation inside executive slide notes
presentation_table = summary_coefs[['Coef.', 'Std.Err.', 't', 'P>|t|']].copy()
presentation_table.columns = ['Coefficient Weight', 'Standard Error', 't-Statistic', 'p-Value (Significance)']

print("📋 EXECUTIVE STATISTICAL ATTRIBUTION MATRIX (95th Percentile Extreme Peak Model)")
print("=================================================================================")
print(presentation_table.round(4).to_string())
print("=================================================================================")
print("💡 Interpretation Tracker: A p-Value < 0.05 proves the variable statistically drives grid load changes.")
```

```python
# Calculate error residuals from the historical base model run
train_stat_pred = median_model.predict(train_stat)
residuals = train_stat['target_load'] - train_stat_pred

# Run a formal Breusch-Pagan test to evaluate residual variance stability
from statsmodels.stats.diagnostic import het_breuschpagan
exog_vars = sm.add_constant(train_stat[['temp_c', 'CDD', 'HDD', 'hour_sin', 'hour_cos', 'load_lag_24']])
lm_stat, lm_p, f_stat, f_p = het_breuschpagan(residuals, exog_vars)

print("🔍 FORMAL RESIDUAL TELEMETRY DIAGNOSTIC STATUS")
print(f"--------------------------------------------------")
print(f"• Breusch-Pagan LM p-Value: {lm_p:.4e}")
if lm_p < 0.05:
    print("⚠️ Status: Heteroskedasticity Detected. Grid volatility changes dynamically with weather intensity.")
else:
    print("✅ Status: Homoskedasticity Confirmed. Error distributions remain stable across parameters.")
print(f"--------------------------------------------------")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Configure industrial economic cost variables
# Assuming a standard post-crisis baseline market price and a grid deviation penalty fee per MWh
BASE_ELECTRICITY_PRICE_MWH = 65.0  # Average Euro per MWh in Spain
IMBALANCE_PENALTY_PER_MWH = 25.0   # Imbalance fine charged by TSO for scheduling errors

# 2. Build economic ledger dataframe from your advanced evaluation results
econ_df = advanced_eval_df.copy()

# Calculate absolute errors in Megawatts
econ_df['lgb_error_mw'] = np.abs(econ_df[target] - econ_df['lgb_pred'])
econ_df['hybrid_error_mw'] = np.abs(econ_df[target] - econ_df['hybrid_ensemble_pred'])

# 3. Compute continuous financial penalties incurred over time
econ_df['LGBM_Financial_Penalty_EUR'] = econ_df['lgb_error_mw'] * IMBALANCE_PENALTY_PER_MWH
econ_df['Hybrid_Financial_Penalty_EUR'] = econ_df['hybrid_error_mw'] * IMBALANCE_PENALTY_PER_MWH

total_lgbm_loss = econ_df['LGBM_Financial_Penalty_EUR'].sum()
total_hybrid_loss = econ_df['Hybrid_Financial_Penalty_EUR'].sum()
net_savings = total_lgbm_loss - total_hybrid_loss

print("💼 INDUSTRIAL FINANCIAL LOSS AUDIT REPORT")
print("==================================================")
print(f"• Total Penalties Incurred by LightGBM Baseline : €{total_lgbm_loss:,.2f}")
print(f"• Total Penalties Incurred by Hybrid Ensemble   : €{total_hybrid_loss:,.2f}")
print(f"--------------------------------------------------")
print(f"• 💰 TOTAL CAPITAL RECOVERED VIA HYBRID MODEL   : €{net_savings:,.2f}")
print("==================================================")
print(f"🔥 Optimization Metric: Reduced grid penalty liabilities by {((total_lgbm_loss - total_hybrid_loss)/total_lgbm_loss)*100:.1f}%!")
```

```python
# Create an operational schedule for a flexible industrial facility
econ_df['Optimized_Load_Schedule'] = econ_df['hybrid_ensemble_pred'].copy()

# Identify the highest 3 hours of demand and the lowest 3 hours of demand
peak_hours = econ_df['hybrid_ensemble_pred'].nlargest(3).index
off_peak_hours = econ_df['hybrid_ensemble_pred'].nsmallest(3).index

# Shift 15% of flexible industrial equipment consumption away from peak hours to off-peak slots
flexible_load_pct = 0.15

for ph, oph in zip(peak_hours, off_peak_hours):
    shuffled_load = econ_df.loc[ph, 'hybrid_ensemble_pred'] * flexible_load_pct
    econ_df.loc[ph, 'Optimized_Load_Schedule'] -= shuffled_load
    econ_df.loc[oph, 'Optimized_Load_Schedule'] += shuffled_load

# Calculate total operational energy procurement bills
econ_df['Standard_Bill_EUR'] = econ_df[target] * BASE_ELECTRICITY_PRICE_MWH
econ_df['Optimized_Bill_EUR'] = econ_df['Optimized_Load_Schedule'] * BASE_ELECTRICITY_PRICE_MWH

procure_savings = econ_df['Standard_Bill_EUR'].sum() - econ_df['Optimized_Bill_EUR'].sum()
print(f"⚡ Smart Scheduling Dividend: Shifting flexible capacity saved an additional €{procure_savings:,.2f} in raw utility bills!")
```
